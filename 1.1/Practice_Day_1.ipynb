```python
#LISTS
my_list = [10, 20, 30, 40, 50]
print("First element:", my_list[0])
print("Last element:", my_list[-1])

print("Elements from index 1 to 3 (not inclusive):", my_list[1:4])
print("From start to index 2:", my_list[:3])
print("From index 2 to end:", my_list[2:])
print("Every second element:", my_list[::2])
print("Reversed list:", my_list[::-1])
```

```python
#TUPLES
my_tuple = ("a", "b", "c", "d", "e")
print("Element at index 0:", my_tuple[0])
print("Element at index -2:", my_tuple[-2])

print("Slice from index 1 to 3:", my_tuple[1:4])
print("Tuple reversed:", my_tuple[::-1])
```

```python
#STRINGS
text = "Hello World"
print("Character at index 0:", text[0])
print("Last character:", text[-1])

print("First 6 characters:", text[:6])
print("From 6 to end:", text[6:])
print("Every 2nd character:", text[::2])
print("Reversed string:", text[::-1])
```

```python
#LIST
fruits = ["Apple", "Banana", "Cherry", "Apple"]

print("Original list:", fruits)
print("First fruit:", fruits[0])

fruits.append("Orange")
print("After appending:", fruits)

fruits.remove('Banana')
print("After removing Banana:", fruits)

print("Number of fruits:", len(fruits))
```

```python
#SET
unique_numbers = {1, 2, 3, 2, 1}
print("Set (no duplicates):", unique_numbers)
unique_numbers.add(4)
print("After adding 4:", unique_numbers)
unique_numbers.remove(2)
print("After removing 2:", unique_numbers)
set_a = {1, 2, 3}
set_b = {3, 4, 5}
print("Set a:", set_a)
print("Set b:", set_b)
print("Union:", set_a | set_b)
print("Intersection:", set_a & set_b)
print("Difference (A - B):", set_a - set_b)
```

```python
#DICT
student = {"Name": "Dhanush","Age": 19,"course": "Python"}
print("Dictionary:", student)
print("Student name:", student['Name'])

student["Grade"] = "A"
print("After adding grade:", student)

del student['Age']
print("After deleting age:", student)

for key, value in student.items():
    print(f"{key}: {value}")

#Dictionary from two list
keys = ["Id","Name","Marks"]
values = [101,"Dhanush",94]
student_info = dict(zip(keys, values))
print("Student Info from zip:", student_info)
```

```python
#FUNCTIONS
def greet():
    print("Hello! Welcome to Python Functions.")
greet()
#function with arguments
def greet_user(name):
    print(f"Hello, {name}!")
greet_user("Dhanush")
greet_user("Umar")
greet_user("Sai Ram")
greet_user("Thrishanth")
#Function with default argument
def greet_user(name="Guest"):
    print(f"Welcome, {name}!")
greet_user()
greet_user("Ravi")
```

```python
def print_names(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)

print_names("Dhanush", "Sai Ram", "Umar", "Thrishanth")
```

```python
def print_info(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")

print_info(name="Dhanush", age=19, city="Hyderabad")
```

```python
def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

show_details(1, 2, 3, name="Dhanush", city="Hyderabad")
```

```python
def multiply(a, b):
    return a * b

def square(n):
    return multiply(n, n)

print("Square of 4:", square(4))
```

```python
from datetime import datetime,date,timedelta
now=datetime.now()
print("Now:",now)
today=date.today()
print("Today's Date:",today)
dob=date(2007,10,18)
print("DOB:",dob)
yesterday=today-timedelta(days=1)
tomorrow=today+timedelta(days=1)
print("Yesterday:",yesterday)
print("Tomorrow:",tomorrow)
formatted=now.strftime("%d-%m-%Y %H:%M:%S")
print("Formatted DateTime:",formatted)
```

```python
from collections import Counter
fruits=['apple','banana','apple','orange','banana','apple']
count=Counter(fruits)
print("Fruit Count:",count)
```

```python
from collections import defaultdict
marks=defaultdict(lambda:0)
marks['Math']=90
marks['Science']=80
print("English Marks (not set):",marks['English'])
```

```python
from collections import namedtuple
Point=namedtuple('Point','x y')
p1=Point(10,20)
print("Point:",p1)
print("x:",p1.x,"y:",p1.y)
```

```python
from collections import deque
dq=deque([1,2,3])
dq.append(4)
dq.appendleft(0)
dq.pop()
dq.popleft()
print("Deque after operations:",dq)
```
